```python
import pandas as pd
import numpy as np
import glob
import os
from pathlib import Path
```

```python
file_path = '../Early_embryo/Early_percentages.csv'

df_early = pd.read_csv('../Early_embryo/Early_percentages_per_embryo.csv').drop(columns = 'Unnamed: 0')
df_early
```

```python
file_path = 'Late_embryo/Early_percentages.csv'

df_late = pd.read_csv('../Late_embryo/Late_percentages_per_embryo.csv').drop(columns = 'Unnamed: 0')
df_late
```

```python
early_counts = df_early["Counts_of_not_touching"]
late_counts = df_late["Counts_of_not_touching"]

combined = pd.concat([early_counts, late_counts],axis=1, 
                            keys=['Early_embryo','Late_embryo'])
combined
```

```python
from scipy.stats import mannwhitneyu
```

```python
statistic, p_value = mannwhitneyu(combined["Early_embryo"],
                                  combined["Late_embryo"])
```

```python
print("Mann-Whitney U statistic:", statistic)
print("p-value:", p_value)
```

```python
df_early_dropped = df_early.drop("Embryo_ID", axis=1)
df_late_dropped = df_late.drop("Embryo_ID", axis=1)
```

```python
for column in df_early_dropped.columns:
    early_counts = df_early_dropped[column]
    late_counts = df_late_dropped[column]
    
    combined = pd.concat([early_counts,late_counts],
                         axis=1, 
                         keys=['Early_embryo',
                               'Late_embryo'])
    statistic, p_value = mannwhitneyu(combined
                                      ["Early_embryo"],
                                      combined
                                      ["Late_embryo"])
    print("Column:",column)
    print("Mann-Whitney U statistic:", statistic)
    print("p-value:", p_value)
```

```python
from scipy.stats import combine_pvalues
```

```python
p_values_early_late = [0.32615054791036024,0.483547415282473,0.6694494696186544,0.41322769826984884,
            0.730274307934341,0.31029322753991717,0.6959977439892062,0.10893299764725162]
combined_p_value_early_late =combine_pvalues(p_values_early_late,method='fisher')
```

```python
print (combined_p_value_early_late[1])
```

```python
file_path = '../PPE_late/PPE_percentages_per_embryo.csv'

df_PPE = pd.read_csv('../PPE_late/PPE_percentages_per_embryo.csv').drop(columns = 'Unnamed: 0')
df_PPE
```

```python
df_PPE_dropped = df_PPE.drop("Embryo_ID", axis=1)
df_late_dropped = df_late.drop("Embryo_ID", axis=1)
```

```python
combined_results = pd.DataFrame()

for column in df_PPE_dropped.columns:
    PPE_counts = df_PPE_dropped[column]
    late_counts = df_late_dropped[column]
    combined_PPE = pd.concat([PPE_counts, late_counts], 
                              axis =1,
                              keys = ['PPE', 'Late'])
    statistic, p_value = mannwhitneyu(combined_PPE["PPE"], 
                                      combined_PPE["Late"])
    combined_results[column] = pd.Series({"Mann-Whitney U statistic": statistic, 
                                          "p-value":p_value})
    print("Column:", column)
    print("Mann-Whitney U statistic:", statistic)
    print("p-value:", p_value)
    print()
```
